```python

from sklearn.preprocessing import StandardScaler
cols = ['reading score','writing score','total_score','average_score']
scaler = StandardScaler()

cleaned_df[cols] = scaler.fit_transform(cleaned_df[cols])
```

```python

from scipy.stats import pearsonr

# ----------------------------------
# Pearson Correlation Calculation
# ----------------------------------

# List of features to check against target
selected_features = [
    'gender', 'reading score', 'writing score', 'total_score', 'average_score',
    'lunch', 'test preparation course', 'race/ethnicity_group D',
    'race/ethnicity_group C', 'race/ethnicity_group B', 'race/ethnicity_group E',"parental level of education_bachelor's degree","parental level of education_high school","parental level of education_master's degree","parental level of education_some college","parental level of education_some high school"
]

correlations = {
    feature: pearsonr(cleaned_df[feature], cleaned_df['math score'])[0]
    for feature in selected_features
}
correlation_df = pd.DataFrame(list(correlations.items()), columns=['Feature', 'Pearson Correlation'])
correlation_df.sort_values(by='Pearson Correlation', ascending=False)
```

```python

cat_features = [
    'gender',
    'lunch',
    'test preparation course',
    'race/ethnicity_group B',
    'race/ethnicity_group C',
    'race/ethnicity_group D',
    'race/ethnicity_group E',
    "parental level of education_bachelor's degree",
    'parental level of education_high school',
    "parental level of education_master's degree",
    'parental level of education_some college',
    'parental level of education_some high school'
]
```

```python

from scipy.stats import chi2_contingency
import pandas as pd

alpha = 0.05

cleaned_df['math score_bin'] = pd.qcut(cleaned_df['math score'], q=4, labels=False)
chi2_results = {}

for col in cat_features:
    contingency = pd.crosstab(cleaned_df[col], cleaned_df['math score_bin'])
    chi2_stat, p_val, _, _ = chi2_contingency(contingency)
    decision = 'Reject Null (Keep Feature)' if p_val < alpha else 'Accept Null (Drop Feature)'
    chi2_results[col] = {
        'chi2_statistic': chi2_stat,
        'p_value': p_val,
        'Decision': decision
    }
    chi2_df = pd.DataFrame(chi2_results).T
chi2_df = chi2_df.sort_values(by='p_value')
chi2_df
```

```python
final_df = cleaned_df[['gender', 'lunch', 'test preparation course', 'math score',
       'reading score', 'writing score', 'total_score', 'average_score',
       'race/ethnicity_group E', 'parental level of education_high school',
       "parental level of education_master's degree", 'parental level of education_some college']]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
y=final_df['math score']
```

```python
X = final_df.drop('math score', axis=1)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
y_test
```

```python
r2=r2_score(y_test,y_pred)
```

```python
r2
```

```python
n = X_test.shape[0]
p = X_test.shape[1]
adjusted_r2 = 1 - ((1 - r2) * (n - 1)) / (n - p - 1)
print("Adjusted R² Score:", adjusted_r2)
```
